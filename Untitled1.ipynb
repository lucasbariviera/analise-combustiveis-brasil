```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo dos gráficos
sns.set_theme(style="whitegrid")

# 1. CARREGAMENTO DOS DADOS
ficheiros = [
    "Preços semestrais - AUTOMOTIVOS_2025.01.csv",
    "Preços semestrais - AUTOMOTIVOS_2025.02.csv",
    "Preços semestrais - AUTOMOTIVOS_2026.01.csv"
]

lista_dfs = []
for ficheiro in ficheiros:
    df_temp = pd.read_csv(ficheiro, sep=';', encoding='latin1')
    lista_dfs.append(df_temp)

# Unir todos os ficheiros num único DataFrame
df = pd.concat(lista_dfs, ignore_index=True)

# 2. LIMPEZA E TRATAMENTO DE DADOS
# CORREÇÃO: Usar a posição da coluna (0 e 1) para evitar erros com carateres ocultos
df.rename(columns={
    df.columns[0]: 'Regiao',
    df.columns[1]: 'Estado',
    'Valor de Venda': 'Preco_Venda'
}, inplace=True)

df['Preco_Venda'] = df['Preco_Venda'].astype(str).str.replace(',', '.').astype(float)
df['Data da Coleta'] = pd.to_datetime(df['Data da Coleta'], format='%d/%m/%Y', errors='coerce')
df['Ano'] = df['Data da Coleta'].dt.year
df['Mes'] = df['Data da Coleta'].dt.month
df.dropna(subset=['Preco_Venda', 'Ano'], inplace=True)

# 3. ANÁLISE EXPLORATÓRIA
print("--- Resumo Estatístico dos Preços por Produto ---")
display(df.groupby(['Ano', 'Produto'])['Preco_Venda'].describe())


# 4. VISUALIZAÇÕES MELHORADAS
# Fazer uma cópia explícita para evitar avisos do Pandas
gasolina_df = df[df['Produto'].str.contains('GASOLINA', na=False)].copy()

# --- MELHORIA GRÁFICO 1: Agrupar por mês para suavizar a linha ---
# Criar uma coluna que representa apenas o Ano e o Mês
gasolina_df['Ano_Mes'] = gasolina_df['Data da Coleta'].dt.to_period('M')

# Calcular a média de preço por Mês e por Região
df_mensal = gasolina_df.groupby(['Ano_Mes', 'Regiao'])['Preco_Venda'].mean().reset_index()

# Converter de volta para data (para o gráfico entender o eixo X)
df_mensal['Data'] = df_mensal['Ano_Mes'].dt.to_timestamp()

plt.figure(figsize=(12, 6))
# Adicionado marker='o' para colocar uma bolinha em cada mês
sns.lineplot(x='Data', y='Preco_Venda', hue='Regiao', data=df_mensal, marker='o', linewidth=2)
plt.title('Evolução do Preço Médio Mensal da Gasolina por Região (2025 - 2026)', fontsize=15, fontweight='bold')
plt.ylabel('Preço Médio (R$)', fontsize=12)
plt.xlabel('Mês', fontsize=12)
plt.legend(title='Região', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7) # Deixa a grelha de fundo mais suave
plt.tight_layout()
plt.savefig('evolucao_temporal_gasolina_melhorada.png', dpi=300) # Salva com melhor resolução
plt.show()


# --- MELHORIA GRÁFICO 2: Ordenar as caixas (Boxplot) ---
top_bandeiras = gasolina_df['Bandeira'].value_counts().head(5).index
df_top_bandeiras = gasolina_df[gasolina_df['Bandeira'].isin(top_bandeiras)]

# Ordenar as bandeiras do maior preço médio para o menor
ordem_bandeiras = df_top_bandeiras.groupby('Bandeira')['Preco_Venda'].median().sort_values(ascending=False).index

plt.figure(figsize=(14, 7))
sns.boxplot(x='Bandeira', y='Preco_Venda', hue='Ano', data=df_top_bandeiras, order=ordem_bandeiras, palette='Set2')
plt.title('Distribuição do Preço da Gasolina nas Principais Bandeiras (2025 vs 2026)', fontsize=15, fontweight='bold')
plt.ylabel('Preço (R$)', fontsize=12)
plt.xlabel('Bandeira', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('boxplot_bandeiras_anos_melhorado.png', dpi=300)
plt.show()
```
